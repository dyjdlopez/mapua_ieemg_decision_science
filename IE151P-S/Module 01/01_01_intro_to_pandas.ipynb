### What is pandas?

**[pandas](https://pandas.pydata.org/docs/)** is an open-source Python library for working with tabular data — the kind of data that naturally sits in rows and columns, like a spreadsheet or a database table. It's built on top of **[NumPy](https://numpy.org/doc/stable/)**, Python's core library for fast numerical computation, and adds the row/column/index structure that raw NumPy arrays don't have on their own.

A useful analogy: think of pandas as Excel's engine, running inside Python. Where a spreadsheet gives you rows, columns, and headers instead of a wall of numbers, pandas gives you the same thing as a Python object you can filter, summarize, and reshape with code — and unlike a spreadsheet, every step is scriptable, repeatable, and works the same way on 10 rows or 10 million.

**A bit of history.** pandas was created in 2008 by Wes McKinney, then a quantitative analyst at AQR Capital Management, who needed a tool as flexible as R's `data.frame` for handling financial time-series data in Python — at the time, Python had good tools for basic data preparation, but very little for structured data analysis. He released the library publicly in 2009, and it has since become the standard tool for tabular data across data science, finance, and research.

**Where does the name come from?** Despite the panda-bear logo, pandas isn't named after the animal. According to McKinney's own account, the name is derived from **panel data** — an econometrics term for multidimensional datasets that track multiple subjects over multiple time periods — combined with a play on "**P**ython **Da**ta A**n**alysi**s**." The panda-bear branding came afterward, because the shortened name happened to evoke it.

Two building blocks do almost everything in pandas:

- **[Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html)** — a single labeled column of data (think: one column of a spreadsheet, plus the row labels next to it)
- **[DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html)** — a full table made of one or more Series that all share the same row labels (the **index**)

Every dataset you'll touch for the rest of this course — the messy patient survey table from Lesson 1.3, and the flight delay data we'll use going forward — gets loaded and manipulated through pandas first, before any cleaning, visualization, or modeling happens. Everything else in this notebook is really just: *how do I build these Series and DataFrames, look inside them, and pull out the piece I need?*


## 2. Worked Example

### 2.1 The Pandas Library

Before anything else, we import pandas. The `as pd` part is just a nickname — it's the near-universal convention, so every pandas tutorial and every piece of code you copy from a colleague will assume `pd` means pandas.


```python
import pandas as pd

print("pandas version:", pd.__version__)
```

### 2.2 Primary Data Type: [Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html)

A **[Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html)** is one column of data, plus a label for every row — that row label is called the **[index](https://pandas.pydata.org/docs/reference/api/pandas.Series.index.html)**. Let's build one by hand: defect counts for five production batches.


```python
defect_counts = pd.Series(
    [3, 5, 2, 4, 6],
    index=['B001', 'B002', 'B003', 'B004', 'B005'],
    name='defect_count'
)

defect_counts
```

Notice three things printed above: the **[index](https://pandas.pydata.org/docs/reference/api/pandas.Series.index.html)** (B001–B005, on the left), the **[values](https://pandas.pydata.org/docs/reference/api/pandas.Series.values.html)** (3, 5, 2, 4, 6), and the Series' **name** (`defect_count`). You can pull each of those out on its own:


```python
print("Index:", defect_counts.index.tolist())
print("Values:", defect_counts.values)
```

### 2.3 Primary Data Type: [DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html)

A **[DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html)** is a full table — several Series stacked side by side, all sharing the same index. The easiest way to build one by hand is from a dictionary, where each key becomes a column name.


```python
quality_data = {
    'batch_id': ['B001', 'B002', 'B003', 'B004', 'B005'],
    'units_produced': [120, 135, 128, 110, 142],
    'defect_count': [3, 5, 2, 4, 6],
    'inspector': ['Reyes', 'Cruz', 'Reyes', 'Santos', 'Cruz']
}

quality_df = pd.DataFrame(quality_data)
quality_df
```

This is a small, **ideal-case** dataset on purpose — five rows, no missing values, nothing messy. We'll compare it against a real, messier dataset in a few cells. For now, let's look at its [`.shape`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.shape.html), [`.columns`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.columns.html), and [`.index`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.index.html):


```python
print("Shape (rows, columns):", quality_df.shape)
print("Columns:", quality_df.columns.tolist())
print("Index:", quality_df.index.tolist())
```

### 2.4 Indexing into a DataFrame

There are a few ways to pull data out of a DataFrame, and they answer different questions:

- **Square brackets with one column name** → a single Series
- **Square brackets with a list of column names** → a smaller DataFrame
- **[`.loc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html)** → select by row/column **label**
- **[`.iloc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iloc.html)** → select by row/column **position** (like a plain Python list)


```python
# One column, by name -> a Series
quality_df['units_produced']
```

```python
# Multiple columns, by name -> a DataFrame
quality_df[['batch_id', 'defect_count']]
```

```python
# One row, by label, using .loc
quality_df.loc[0]
```

```python
# The first two rows, by position, using .iloc
quality_df.iloc[0:2]
```

### 2.5 Basic Analysis: [`.describe()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html) — the Ideal Case

`.describe()` gives you a fast statistical summary of every numeric column: how many values there are, their average, spread, and range. Let's run it on our clean toy table first, before we see what it looks like on real (messier) data.


```python
quality_df.describe()
```

Every row here means something specific:

| Row | Meaning |
|---|---|
| `count` | how many non-missing values are in that column |
| `mean` | the average |
| `std` | standard deviation — how spread out the values are |
| `min` / `max` | the smallest and largest values |
| `25%` / `50%` / `75%` | the quartiles — `50%` is the median |

Notice `count` is `5` for both numeric columns — no surprises, because this table has no missing values. Keep that number in mind; it won't stay that simple once we load real data.

Hand-built tables are fine for learning the mechanics, but every real dataset gets loaded from a file. We'll use the **Palmer Penguins** dataset — real measurements of penguins from three islands near Palmer Station, Antarctica. [`pd.read_csv()`](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html) can load a CSV directly from a URL, which is the easiest way to work with data in Colab.

**Citations:**

```python
penguins_url = "https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv"
penguins_df = pd.read_csv(penguins_url)

penguins_df.head()
```

[`.head()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.head.html) shows the first 5 rows by default — useful for a quick sanity check without printing the whole table. [`.shape`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.shape.html) and [`.info()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.info.html) give you the bigger picture: how big the table is, and what type of data lives in each column.


```python
print("Shape (rows, columns):", penguins_df.shape)
penguins_df.info()
```

### 2.7 Basic Analysis: [`.describe()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html) — the Real Case

Now let's run the same `.describe()` we ran on the toy table — but on real data.


```python
penguins_df.describe()
```

Compare this to Section 2.5. The dataset has **344 rows** total (from `.shape` above), but look at the `count` row here — every numeric column shows fewer than 344. That gap between the row count and the `count` in `.describe()` is your first clue that a column has **missing values**.

This is exactly the kind of signal we'll act on in Notebook 01-02, when we get into cleaning techniques properly. For now, just notice: **`.describe()` on real data doesn't just summarize — it can also reveal a problem.**


### 2.8 Selecting Rows and Columns (Slicing)

The same selection patterns from Section 2.4 apply here, just on a bigger, real table.


```python
# One column, by name -> a Series
penguins_df['species'].head()
```

```python
# Multiple columns, by name -> a DataFrame
penguins_df[['species', 'island', 'body_mass_g']].head()
```

```python
# First 5 rows, by position, using .iloc
penguins_df.iloc[0:5]
```

[`.loc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html) also accepts a **condition** instead of a label — this is how you filter rows based on a value, not just a position.


```python
# Rows where species is "Adelie"
penguins_df.loc[penguins_df['species'] == 'Adelie'].head()
```

You can combine row and column selection in a single [`.loc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html) call: rows first, columns second, separated by a comma.


```python
# Rows 10 through 20, but only the "island" and "flipper_length_mm" columns
penguins_df.loc[10:20, ['island', 'flipper_length_mm']]
```

## 5. Wrap-Up

- **pandas** turns tabular data into something Python can filter, summarize, and reshape — the same rows/columns/index structure as a spreadsheet, but scriptable
- A **Series** is one labeled column; a **DataFrame** is a table made of Series sharing an index
- `.describe()` is a fast way to see the shape of your numeric data — and a mismatch between `count` and your row total is often the first sign something needs cleaning
- `pd.read_csv()` loads real data directly from a URL, no manual download needed in Colab
- `.loc[]` (by label/condition) and `.iloc[]` (by position) are how you slice out exactly the rows and columns you need
